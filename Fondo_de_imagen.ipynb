```python
import os
import shutil
from PIL import Image, ImageChops

# 1. DEFINICIÓN DE RUTAS (CON COMILLAS)
archivo_zip_entrada = "/content/Prueba de imagenes.zip"
carpeta_extraccion = "/content/imagenes_temp"
carpeta_procesadas = "/content/resultado_final"
nombre_zip_salida = "Resultado_Final_v3" # Se le agregará .zip automáticamente

umbral = 20

# --- PASO A: LIMPIEZA Y EXTRACCIÓN ---
if os.path.exists(carpeta_extraccion): shutil.rmtree(carpeta_extraccion)
if os.path.exists(carpeta_procesadas): shutil.rmtree(carpeta_procesadas)

os.makedirs(carpeta_procesadas, exist_ok=True)

if os.path.exists(archivo_zip_entrada):
    !unzip -q "{archivo_zip_entrada}" -d "{carpeta_extraccion}"
    print("✅ ZIP descomprimido correctamente.")
else:
    print(f"❌ No se encontró el archivo: {archivo_zip_entrada}")

# --- PASO B: PROCESAMIENTO ---
for raiz, carpetas, archivos in os.walk(carpeta_extraccion):
    for nombre_archivo in archivos:
        if nombre_archivo.lower().endswith((".jpg", ".png", ".jpeg")) and not nombre_archivo.startswith("._"):
            try:
                img_path = os.path.join(raiz, nombre_archivo)
                img = Image.open(img_path).convert("RGBA")

                # Detectar bordes comparando con blanco
                gris = img.convert("L")
                foto_bw = gris.point(lambda x: 0 if x < (255 - umbral) else 255)
                foto_invertida = ImageChops.invert(foto_bw)

                bbox = foto_invertida.getbbox()

                if bbox:
                    img_recortada = img.crop(bbox)

                    # Guardar como PNG para mantener calidad
                    nombre_base = os.path.splitext(nombre_archivo)[0]
                    destino = os.path.join(carpeta_procesadas, f"{nombre_base}.png")
                    img_recortada.save(destino)
                    print(f"✅ Recortada: {nombre_archivo}")
                else:
                    print(f"❌ No se detectó objeto en: {nombre_archivo}")

            except Exception as e:
                print(f"⚠️ Error en {nombre_archivo}: {e}")

# --- PASO C: COMPRESIÓN FINAL ---
shutil.make_archive(nombre_zip_salida, 'zip', carpeta_procesadas)
print(f"\n🔥 PROCESO TERMINADO. Descarga '{nombre_zip_salida}.zip' desde el panel izquierdo.")
```
